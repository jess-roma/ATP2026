### Exercício 1: Criar uma lista 
- Defina uma função que recebe um argumento **N** e devolve uma  lista com os números ímpares até **N**;
- Crie um programa que chama a função e imprime a lista resultante.

```python
# Solução do exercício 1
def criaLista(N):
    lista=[]
    contador=0
    while contador<N:
        if contador%2!=0:
            lista.append(contador)
        contador+=1
    return lista
N = int(input("Insira um número:"))
print(criaLista(N))

def criaLista2(N):
    lista=[]
    for elem in range(N):
        if elem%2!=0:
            lista.append(elem)
        return lista

def criaLista(N):
    lista=[]
    for elem in range(1,N,2):
        lista.append(elem)
    return lista
print(criaLista(N))
```

### Exercício 4: Quadrado dos elementos de uma lista

- Defina uma função que recebe uma lista como argumento e devolve uma nova lista com os quadrados de cada número da lista inicial. 

#### Input (exemplo)
in: [1, 2, 3, 4]
#### Output (exemplo)
out: [1, 4, 9, 16]

```python
# Solução do exercício 4
def quadrados(lista):
    listaq=[]
    for elem in lista:
        listaq.append(elem*elem)
    return listaq
lista= [1, 2, 3, 4]
print(quadrados(lista))
```

```python
# Lista em compreensão 
# [expressao for elem in lista if condicao]

listaA= [1,2,3,4]

listaB= [elem**2 for elem in listaA]
print(listaB)

listaC= [elem**2 for elem in listaA if elem%2==0]
print(listaC)

def quadrados(lista):
    return [elem**2 for elem in listaA]
```

### Exercício 6: Menor de uma sequência
- Defina uma função que recebe uma lista como argumento e devolve o menor número nessa lista.
